# Monitor runtime

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import glob
```

```python
df = pd.DataFrame()
all_files = glob.glob('*_time_run.csv')
for filepath in all_files: 
    match = re.search(r'_(\d+)n_', filepath)
    n = int(match.group(1))
    df_load = pd.read_csv(filepath).assign(n=n)
    df = pd.concat([df, df_load])
df
```

```python
plt.figure()
classes = df['n'].unique()
#plt.hist(df[df['n'] == 20][['time_generation', 'time_simulation']], label=['time generation', 'time simulation'])
plt.hist(df[['time_generation', 'time_simulation']], label=['time generation', 'time simulation'])
plt.legend()
plt.show()
```
